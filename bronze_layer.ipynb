#Bronze layer

```python
df_trips = spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load("/Volumes/demo/demo/taxivolume/taxi_trip_data_raw.csv")

df_trips.show(5)
```

```python
df_trips.write.format("delta").mode("overwrite").saveAsTable("demo.demo.bronze_taxi_trips")
```
