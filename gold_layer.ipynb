#Gold layer

```python
%sql
create or replace table demo.demo.gold_zone_revenue as
select
pickup_zone,
sum(total_amount) as total_revenue,
count(*) as total_trips
from demo.demo.silver_taxi_trips_enriched
group by pickup_zone
order by total_revenue desc
```

```python
%sql
select 
pickup_zone,
pickup_latitude,
pickup_longitude,
SUM(total_amount) as total_revenue,
count(*) as total_trips
from demo.demo.silver_taxi_trips_enriched
group by pickup_zone, pickup_latitude, pickup_longitude
limit 10;
```

#Peak hours (clean & correct)

```python
%sql
create or replace table demo.demo.gold_peak_hours as
select
  pickup_hour,
  count(*) as total_trips,
  sum(total_amount) as total_revenue
from demo.demo.silver_taxi_trips_enriched
group by pickup_hour
order by pickup_hour;
```

#Pickup vs Dropoff time analysis

```python
%sql
Create or replace table demo.demo.gold_time_of_day_revenue as
select
 case
    when pickup_hour between 6 and 11 then 'morning'
    when pickup_hour between 12 and 17 then 'afternoon'
    when pickup_hour between 18 and 23 then 'evening'
    else 'midnight'
    END AS time_of_day,
    sum(total_amount) as total_revenue,
    count(*) as total_trips
from demo.demo.silver_taxi_trips_enriched
group by time_of_day
order by total_revenue desc
```
